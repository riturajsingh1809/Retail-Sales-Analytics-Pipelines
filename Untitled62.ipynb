```python
import pandas as pd
import numpy as np

# --------------------------------------------------
# 1. Read Cleaned Dataset
# --------------------------------------------------

df = pd.read_excel("Retail_Sales_Cleaned.xlsx")

print("Dataset Shape:", df.shape)


# --------------------------------------------------
# 2. Basic Dataset Information
# --------------------------------------------------

print("\nDataset Information:")
df.info()


# --------------------------------------------------
# 3. Descriptive Statistics
# --------------------------------------------------

print("\nDescriptive Statistics:")
display(df.describe())


# --------------------------------------------------
# 4. Total Sales
# --------------------------------------------------

total_sales = df["Sales"].sum()


# --------------------------------------------------
# 5. Total Profit
# --------------------------------------------------

total_profit = df["Profit"].sum()


# --------------------------------------------------
# 6. Total Orders
# --------------------------------------------------

total_orders = df["Order_ID"].nunique()


# --------------------------------------------------
# 7. Total Quantity
# --------------------------------------------------

total_quantity = df["Quantity"].sum()


# --------------------------------------------------
# 8. Average Sales
# --------------------------------------------------

average_sales = df["Sales"].mean()


# --------------------------------------------------
# 9. Average Profit
# --------------------------------------------------

average_profit = df["Profit"].mean()


# --------------------------------------------------
# 10. Overall KPI Analysis
# --------------------------------------------------

kpi = pd.DataFrame({
    "Metric": [
        "Total Sales",
        "Total Profit",
        "Total Orders",
        "Total Quantity",
        "Average Sales",
        "Average Profit"
    ],
    "Value": [
        total_sales,
        total_profit,
        total_orders,
        total_quantity,
        average_sales,
        average_profit
    ]
})

print("\nOverall KPI Analysis:")
display(kpi)


# --------------------------------------------------
# 11. Category-wise Analysis
# --------------------------------------------------

category_analysis = (
    df.groupby("Category")
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Total_Quantity=("Quantity", "sum"),
        Total_Orders=("Order_ID", "nunique")
    )
    .reset_index()
)

print("\nCategory-wise Analysis:")
display(category_analysis)


# --------------------------------------------------
# 12. Region-wise Analysis
# --------------------------------------------------

region_analysis = (
    df.groupby("Region")
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Total_Orders=("Order_ID", "nunique")
    )
    .reset_index()
)

print("\nRegion-wise Analysis:")
display(region_analysis)


# --------------------------------------------------
# 13. Product-wise Analysis
# --------------------------------------------------

product_analysis = (
    df.groupby("Product")
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Total_Quantity=("Quantity", "sum"),
        Total_Orders=("Order_ID", "nunique")
    )
    .reset_index()
    .sort_values(
        "Total_Sales",
        ascending=False
    )
)

print("\nProduct-wise Analysis:")
display(product_analysis)


# --------------------------------------------------
# 14. Payment Mode Analysis
# --------------------------------------------------

payment_analysis = (
    df.groupby("Payment_Mode")
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Orders=("Order_ID", "nunique")
    )
    .reset_index()
)

print("\nPayment Mode Analysis:")
display(payment_analysis)


# --------------------------------------------------
# 15. Monthly Analysis
# --------------------------------------------------

df["Month"] = df["Order_Date"].dt.month
df["Month_Name"] = df["Order_Date"].dt.month_name()

monthly_analysis = (
    df.groupby(
        ["Month", "Month_Name"]
    )
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Total_Orders=("Order_ID", "nunique")
    )
    .reset_index()
    .sort_values("Month")
)

print("\nMonthly Analysis:")
display(monthly_analysis)


# --------------------------------------------------
# 16. Profit Margin Analysis
# --------------------------------------------------

total_profit_margin = (
    df["Profit"].sum() /
    df["Sales"].sum()
) * 100

profit_margin = pd.DataFrame({
    "Metric": ["Overall Profit Margin"],
    "Value": [total_profit_margin]
})

print("\nProfit Margin:")
display(profit_margin)


# --------------------------------------------------
# 17. Highest Sales Category
# --------------------------------------------------

highest_category = category_analysis.loc[
    category_analysis["Total_Sales"].idxmax(),
    "Category"
]

print("\nHighest Sales Category:")
print(highest_category)


# --------------------------------------------------
# 18. Highest Sales Region
# --------------------------------------------------

highest_region = region_analysis.loc[
    region_analysis["Total_Sales"].idxmax(),
    "Region"
]

print("\nHighest Sales Region:")
print(highest_region)


# --------------------------------------------------
# 19. Best Selling Product
# --------------------------------------------------

best_product = product_analysis.iloc[0]["Product"]

print("\nBest Selling Product:")
print(best_product)


# --------------------------------------------------
# 20. Create ONE Final Analysis Output File
# --------------------------------------------------

with pd.ExcelWriter(
    "Retail_Sales_Analysis.xlsx",
    engine="openpyxl"
) as writer:

    kpi.to_excel(
        writer,
        sheet_name="KPI",
        index=False
    )

    category_analysis.to_excel(
        writer,
        sheet_name="Category_Analysis",
        index=False
    )

    region_analysis.to_excel(
        writer,
        sheet_name="Region_Analysis",
        index=False
    )

    product_analysis.to_excel(
        writer,
        sheet_name="Product_Analysis",
        index=False
    )

    payment_analysis.to_excel(
        writer,
        sheet_name="Payment_Analysis",
        index=False
    )

    monthly_analysis.to_excel(
        writer,
        sheet_name="Monthly_Analysis",
        index=False
    )

    profit_margin.to_excel(
        writer,
        sheet_name="Profit_Margin",
        index=False
    )


# --------------------------------------------------
# 21. Final Message
# --------------------------------------------------

print("\nAnalysis completed successfully!")
print("Final output file: Retail_Sales_Analysis.xlsx")
```
